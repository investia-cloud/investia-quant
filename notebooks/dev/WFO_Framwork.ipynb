```python
def analyze_exposure_regime(
    pf,
    title: str = "Exposure Regime Analysis",
    show_report: bool = True,
    exposure_threshold: float = 0.95,
    flat_threshold: float = 0.05,
    annualization: int = 252,
):
    """
    Analizza 'time-in-market' e comportamento per regime (investito vs flat)
    usando SOLO il Portfolio vectorbt (pf).

    Requisiti minimi su pf:
    - pf.value() (serie equity)
    - pf.gross_exposure() (serie 0..~1)  [in vectorbt esiste; altrimenti fallisce]
      In alternativa, se hai versioni diverse, puoi sostituire con pf.stats('gross_exposure') etc.

    Output:
    - df_summary: tabella con metriche globali + per-regime
    - df_yearly: tabella per anno (% giorni investiti, return investito/flat)
    - (opzionale) stampa con my_display se show_report=True

    Regimi:
    - Invested: gross_exposure >= exposure_threshold
    - Flat: gross_exposure <= flat_threshold
    - Partial: il resto (utile per multi-asset o cash drag)
    """
    if pf is None:
        raise ValueError("pf is None")

    # --- series base ---
    equity = pf.value()
    if isinstance(equity, pd.DataFrame):
        # group -> prendi la prima colonna se serve
        equity = equity.iloc[:, 0]

    # returns giornalieri portafoglio (equity-based)
    rets = equity.pct_change().replace([np.inf, -np.inf], np.nan).fillna(0.0)

    # gross exposure
    if not hasattr(pf, "gross_exposure"):
        raise AttributeError("pf.gross_exposure() non disponibile nella tua versione vectorbt")
    ge = pf.gross_exposure()
    if isinstance(ge, pd.DataFrame):
        ge = ge.iloc[:, 0]
    ge = ge.reindex(rets.index).fillna(method="ffill").fillna(0.0)

    # --- regime masks ---
    invested = ge >= float(exposure_threshold)
    flat = ge <= float(flat_threshold)
    partial = ~(invested | flat)

    def _ann_factor(mask: pd.Series) -> float:
        # annualization effective based on share of time
        p = mask.mean()
        return annualization * p if p > 0 else np.nan

    def _cagr_from_rets(r: pd.Series) -> float:
        if r.empty:
            return np.nan
        cum = (1.0 + r).prod()
        n = len(r)
        if n <= 1:
            return np.nan
        years = n / annualization
        return cum ** (1 / years) - 1 if years > 0 else np.nan

    def _sharpe(r: pd.Series) -> float:
        if r.empty:
            return np.nan
        mu = r.mean()
        sd = r.std(ddof=0)
        if sd == 0:
            return np.nan
        return (mu / sd) * np.sqrt(annualization)

    def _max_dd_from_equity(eq: pd.Series) -> float:
        if eq.empty:
            return np.nan
        peak = eq.cummax()
        dd = eq / peak - 1.0
        return dd.min()

    # --- global ---
    total_time = len(rets)
    summary = {}

    # time in regimes
    summary["Days"] = total_time
    summary["Pct Invested (>=thr)"] = invested.mean()
    summary["Pct Flat (<=thr)"] = flat.mean()
    summary["Pct Partial"] = partial.mean()

    # global perf
    summary["Total Return"] = (equity.iloc[-1] / equity.iloc[0]) - 1 if len(equity) > 1 else np.nan
    summary["CAGR"] = _cagr_from_rets(rets)
    summary["Sharpe"] = _sharpe(rets)
    summary["Max Drawdown"] = _max_dd_from_equity(equity)

    # --- per-regime perf ---
    def _regime_block(mask: pd.Series, name: str):
        r = rets[mask]
        eq = equity.copy()
        # equity restricted to mask dates (not "continuous" DD, but regime-only path)
        eq = eq.loc[r.index]
        summary[f"{name} Days"] = int(mask.sum())
        summary[f"{name} Pct Days"] = mask.mean()
        summary[f"{name} Avg Daily Return"] = r.mean()
        summary[f"{name} Total Return (comp)"] = (1.0 + r).prod() - 1.0
        summary[f"{name} CAGR (naive)"] = _cagr_from_rets(r)
        summary[f"{name} Sharpe"] = _sharpe(r)
        summary[f"{name} Max DD (within regime)"] = _max_dd_from_equity(eq)

    _regime_block(invested, "Invested")
    _regime_block(flat, "Flat")
    _regime_block(partial, "Partial")

    df_summary = pd.DataFrame.from_dict(summary, orient="index", columns=["Value"])

    # make it readable
    pct_rows = [i for i in df_summary.index if "Pct" in i or "Return" in i or "CAGR" in i or "Max Drawdown" in i]
    # Don't over-format numerically here; you already have my_display styling in your framework.

    # --- yearly view ---
    years = rets.index.year
    df = pd.DataFrame({
        "ret": rets,
        "ge": ge,
        "invested": invested,
        "flat": flat,
        "partial": partial,
    })
    grp = df.groupby(years)

    def _year_stats(g: pd.DataFrame) -> pd.Series:
        r = g["ret"]
        inv = g["invested"]
        fl = g["flat"]
        part = g["partial"]
        out = {
            "Pct Invested": inv.mean(),
            "Pct Flat": fl.mean(),
            "Pct Partial": part.mean(),
            "Return Total": (1.0 + r).prod() - 1.0,
            "Return Invested": (1.0 + r[inv]).prod() - 1.0 if inv.any() else 0.0,
            "Return Flat": (1.0 + r[fl]).prod() - 1.0 if fl.any() else 0.0,
            "Return Partial": (1.0 + r[part]).prod() - 1.0 if part.any() else 0.0,
        }
        return pd.Series(out)

    df_yearly = grp.apply(_year_stats)
    df_yearly.index.name = "Year"

    if show_report:
        my_display(df_summary, title=title)
        my_display(df_yearly, title=f"{title} — Yearly Breakdown")

    return df_summary, df_yearly
```

```python
def analyze_timing_efficiency(
    pf,
    portfolio_type: str = "auto",   # "auto", "trading", "rotational", "lazy"
    exposure_threshold: float = 0.95,
    flat_threshold: float = 0.05,
    annualization: int = 252,
    show_report: bool = True,
    vbt_plot_width: int | None = 1000,
    fig_height: int = 700
):
    """
    Timing Efficiency Analysis + Grafici (Plotly) per vectorbt.Portfolio `pf`.
    Ritorna un dict con DataFrame e Figure:
      {
        "df_summary": ...,
        "df_yearly": ...,
        "df_bh_vs_pf": ...,
        "df_timing": ...,
        "aux": {...},
        "portfolio_type": ...,
        "figs": [fig_equity, fig_exposure, fig_yearly]
      }
    """

    # ----- helper inner functions (rifatte in modo compatto) -----
    def _get_series_equity(pf):
        eq = pf.value()
        if isinstance(eq, pd.DataFrame):
            return eq.iloc[:, 0].copy()
        return eq.copy()

    def _safe_rets(series):
        r = series.pct_change().replace([np.inf, -np.inf], np.nan).fillna(0.0)
        return r

    def _cagr_from_rets(r):
        if r.empty:
            return np.nan
        cum = (1.0 + r).prod()
        n = len(r)
        years = n / annualization if annualization > 0 else np.nan
        return cum ** (1 / years) - 1 if years > 0 else np.nan

    def _sharpe(r):
        if r.empty:
            return np.nan
        mu = r.mean()
        sd = r.std(ddof=0)
        if sd == 0 or np.isnan(sd):
            return np.nan
        return (mu / sd) * np.sqrt(annualization)

    def _max_dd_from_equity(eq):
        if eq.empty:
            return np.nan
        peak = eq.cummax()
        dd = eq / peak - 1.0
        return dd.min()

    def _get_ge(pf, eq_index):
        if hasattr(pf, "gross_exposure"):
            ge = pf.gross_exposure()
            if isinstance(ge, pd.DataFrame):
                ge = ge.iloc[:, 0]
            ge = ge.reindex(eq_index).ffill().fillna(0.0)
            return ge
        # fallback: infer from assets * prices
        assets = pf.assets()
        close = getattr(pf, "close", None)
        if isinstance(assets, pd.DataFrame) and close is not None:
            try:
                values = assets * close
                tv = values.sum(axis=1)
                ge_approx = values.abs().sum(axis=1) / tv
                return ge_approx.reindex(eq_index).ffill().fillna(0.0)
            except Exception:
                pass
        if isinstance(assets, pd.Series):
            mask = (assets.abs() > 0).astype(float)
            return mask.reindex(eq_index).ffill().fillna(0.0)
        # last fallback: zeros
        return pd.Series(0.0, index=eq_index)

    def _compute_bh_from_initial_assets(pf, eq_index):
        assets = pf.assets()
        close = getattr(pf, "close", None)
        if close is None:
            return None
        try:
            # Normalize close as DataFrame
            if isinstance(close, pd.Series):
                close = close.to_frame(close.name if close.name is not None else "X")
            if isinstance(assets, pd.Series):
                # try match column
                col = assets.name if assets.name in close.columns else close.columns[0]
                qty0 = float(assets.iloc[0])
                bh = close[col] * qty0
                bh = bh.reindex(eq_index).ffill().fillna(method="ffill")
                return bh
            elif isinstance(assets, pd.DataFrame):
                qty0 = assets.iloc[0].fillna(0.0)
                common = [c for c in qty0.index if c in close.columns]
                if not common:
                    common = close.columns.tolist()[:len(qty0)]
                bh = (qty0[common] * close[common]).sum(axis=1)
                return bh.reindex(eq_index).ffill().fillna(method="ffill")
        except Exception:
            return None
        return None

    def _contiguous_segments(mask: pd.Series):
        """Ritorna lista di (start, end) per i tratti True continui su mask (index sono datetimes)."""
        segments = []
        if mask.empty:
            return segments
        is_on = False
        start = None
        for dt, val in mask.items():
            if val and not is_on:
                is_on = True
                start = dt
            elif not val and is_on:
                is_on = False
                segments.append((start, prev_dt))
            prev_dt = dt
        if is_on:
            segments.append((start, prev_dt))
        return segments

    # ----- MAIN -----
    if pf is None:
        raise ValueError("pf is None")

    eq = _get_series_equity(pf).dropna()
    if eq.empty:
        raise ValueError("pf.value() produced empty equity series.")
    rets = _safe_rets(eq)
    ge = _get_ge(pf, eq.index)

    invested_mask = ge >= float(exposure_threshold)
    flat_mask = ge <= float(flat_threshold)
    partial_mask = ~(invested_mask | flat_mask)

    # infer portfolio_type if auto (heuristica semplice)
    if portfolio_type == "auto":
        close = getattr(pf, "close", None)
        assets = pf.assets()
        n_tickers = None
        if close is not None:
            n_tickers = close.shape[1] if isinstance(close, pd.DataFrame) else 1
        else:
            n_tickers = assets.shape[1] if isinstance(assets, pd.DataFrame) else 1
        if n_tickers == 1:
            portfolio_type = "trading"
        else:
            # detect lazy -> holdings constant
            try:
                if isinstance(assets, pd.DataFrame):
                    const_holdings = (assets.diff().abs().sum().sum() == 0)
                    portfolio_type = "lazy" if const_holdings else "rotational"
                else:
                    portfolio_type = "rotational"
            except Exception:
                portfolio_type = "rotational"

    # summary metrics (as before)
    summary = {}
    summary["Days"] = len(rets)
    summary["Pct Invested (>=thr)"] = invested_mask.mean()
    summary["Pct Flat (<=thr)"] = flat_mask.mean()
    summary["Pct Partial"] = partial_mask.mean()
    summary["Total Return"] = (eq.iloc[-1] / eq.iloc[0]) - 1.0
    summary["CAGR"] = _cagr_from_rets(rets)
    summary["Sharpe"] = _sharpe(rets)
    summary["Max Drawdown"] = _max_dd_from_equity(eq)

    def _regime_stats(mask, label):
        r = rets[mask]
        out = {}
        out[f"{label} Days"] = int(mask.sum())
        out[f"{label} Pct Days"] = mask.mean()
        out[f"{label} Total Return (comp)"] = (1.0 + r).prod() - 1.0 if len(r) > 0 else np.nan
        out[f"{label} CAGR"] = _cagr_from_rets(r) if len(r) > 0 else np.nan
        out[f"{label} Sharpe"] = _sharpe(r)
        try:
            eq_slice = eq.loc[r.index]
            out[f"{label} Max DD (within regime)"] = _max_dd_from_equity(eq_slice)
        except Exception:
            out[f"{label} Max DD (within regime)"] = np.nan
        return out

    summary.update(_regime_stats(invested_mask, "Invested"))
    summary.update(_regime_stats(flat_mask, "Flat"))
    summary.update(_regime_stats(partial_mask, "Partial"))

    df_summary = pd.DataFrame.from_dict(summary, orient="index", columns=["Value"])

    # yearly breakdown
    yrs = rets.index.year
    df_work = pd.DataFrame({
        "ret": rets,
        "ge": ge,
        "invested": invested_mask,
        "flat": flat_mask,
        "partial": partial_mask
    })
    def _year_stats(group):
        r = group["ret"]
        inv = group["invested"]
        fl = group["flat"]
        part = group["partial"]
        res = {
            "Pct Invested": inv.mean(),
            "Pct Flat": fl.mean(),
            "Pct Partial": part.mean(),
            "Return Total": (1.0 + r).prod() - 1.0 if len(r) > 0 else np.nan,
            "Return Invested": (1.0 + r[inv]).prod() - 1.0 if inv.any() else 0.0,
            "Return Flat": (1.0 + r[fl]).prod() - 1.0 if fl.any() else 0.0,
            "Return Partial": (1.0 + r[part]).prod() - 1.0 if part.any() else 0.0,
        }
        return pd.Series(res)

    df_yearly = df_work.groupby(yrs).apply(_year_stats)
    df_yearly.index.name = "Year"

    # BH comparator
    bh_series = _compute_bh_from_initial_assets(pf, eq.index)
    bh_cmp = {}
    if bh_series is not None:
        bh_series = bh_series.reindex(eq.index).ffill().fillna(method="ffill")
        bh_rets = _safe_rets(bh_series)
        bh_cmp["BH Total Return"] = (bh_series.iloc[-1] / bh_series.iloc[0]) - 1.0
        bh_cmp["BH CAGR"] = _cagr_from_rets(bh_rets)
        bh_cmp["BH Max Drawdown"] = _max_dd_from_equity(bh_series)
        bh_cmp["Excess CAGR (PF - BH)"] = df_summary.loc["CAGR", "Value"] - bh_cmp["BH CAGR"]
        bh_cmp["Drawdown Avoided (BH DD - PF DD)"] = bh_cmp["BH Max Drawdown"] - df_summary.loc["Max Drawdown", "Value"]
        # BH/PF during invested days
        inv_idx = invested_mask[invested_mask].index
        try:
            bh_r_inv = (1.0 + bh_rets.loc[inv_idx]).prod() - 1.0 if len(inv_idx) > 0 else np.nan
        except Exception:
            bh_r_inv = np.nan
        pf_r_inv = (1.0 + rets.loc[inv_idx]).prod() - 1.0 if len(inv_idx) > 0 else np.nan
        bh_cmp["BH Return During Invested Days"] = bh_r_inv
        bh_cmp["PF Return During Invested Days"] = pf_r_inv
    else:
        bh_cmp = {k: np.nan for k in [
            "BH Total Return","BH CAGR","BH Max Drawdown","Excess CAGR (PF - BH)",
            "Drawdown Avoided (BH DD - PF DD)","BH Return During Invested Days","PF Return During Invested Days"
        ]}

    df_bh_vs_pf = pd.DataFrame.from_dict(bh_cmp, orient="index", columns=["Value"])

    # Timing metrics
    invested_days = int(invested_mask.sum())
    pct_invested = invested_mask.mean()
    pf_cagr = df_summary.loc["CAGR", "Value"]
    timing = {
        "CAGR": pf_cagr,
        "Time Invested (days)": invested_days,
        "Pct Time Invested": pct_invested,
        "CAGR per Invested Year": (pf_cagr / pct_invested) if (pct_invested and not np.isnan(pf_cagr)) else np.nan
    }
    df_timing = pd.DataFrame.from_dict(timing, orient="index", columns=["Value"])

    # ----- GRAFICI Plotly -----
    figs = []

    # 1) Equity vs BH with invested shading
    fig_eq = go.Figure()
    fig_eq.add_trace(go.Scatter(x=eq.index, y=eq.values, mode="lines", name="PF Equity", line=dict(color="#1f77b4")))
    if bh_series is not None:
        fig_eq.add_trace(go.Scatter(x=bh_series.index, y=bh_series.values, mode="lines", name="Buy & Hold", line=dict(color="#ff7f0e", dash="dot")))
    # add invested shaded rectangles
    for (s,e) in _contiguous_segments(invested_mask):
        fig_eq.add_vrect(x0=s, x1=e, fillcolor="green", opacity=0.08, layer="below", line_width=0)
    fig_eq.update_layout(title=f"{pf.name if hasattr(pf,'name') else 'Portfolio'} — Equity vs BH (shaded = Invested)", hovermode="x unified")
    fig_eq.update_xaxes(title_text="Date")
    fig_eq.update_yaxes(title_text="Value")
    if vbt_plot_width is not None:
        fig_eq.update_layout(width=int(vbt_plot_width), height=int(fig_height*0.45))
    figs.append(fig_eq)

    # 2) Gross Exposure with color by regime (stacked background)
    fig_ge = go.Figure()
    fig_ge.add_trace(go.Scatter(x=ge.index, y=ge.values, mode="lines", name="Gross Exposure", line=dict(color="#d62728")))
    # background colored bands for regimes (Invested green, Flat grey, Partial orange)
    # build contiguous segments for each mask
    for (s,e) in _contiguous_segments(invested_mask):
        fig_ge.add_vrect(x0=s, x1=e, fillcolor="green", opacity=0.08, layer="below", line_width=0)
    for (s,e) in _contiguous_segments(flat_mask):
        fig_ge.add_vrect(x0=s, x1=e, fillcolor="lightgrey", opacity=0.06, layer="below", line_width=0)
    for (s,e) in _contiguous_segments(partial_mask):
        fig_ge.add_vrect(x0=s, x1=e, fillcolor="orange", opacity=0.04, layer="below", line_width=0)
    fig_ge.update_layout(title="Gross Exposure (background: Invested/Flat/Partial)", hovermode="x unified")
    fig_ge.update_xaxes(title_text="Date")
    fig_ge.update_yaxes(title_text="Gross Exposure", range=[-0.05, 1.05])
    if vbt_plot_width is not None:
        fig_ge.update_layout(width=int(vbt_plot_width), height=int(fig_height*0.25))
    figs.append(fig_ge)

    # 3) Yearly returns PF vs BH (if BH available)
    years = sorted(df_yearly.index.tolist())
    pf_returns = [df_yearly.loc[y, "Return Total"] if y in df_yearly.index else np.nan for y in years]
    bh_returns = []
    if bh_series is not None:
        bh_rets = (1 + bh_series.pct_change().fillna(0)).cumprod()
        # compute yearly BH returns per same years
        bh_returns = []
        bh_r = (1 + bh_series.pct_change().fillna(0))
        for y in years:
            idx_year = bh_series[bh_series.index.year == y].index
            if len(idx_year) == 0:
                bh_returns.append(np.nan)
            else:
                bh_returns.append((1.0 + bh_r.loc[idx_year]).prod() - 1.0)
    fig_year = go.Figure()
    x = [str(y) for y in years]
    fig_year.add_trace(go.Bar(x=x, y=[100*v for v in pf_returns], name="PF Return (%)", marker_color="#1f77b4"))
    if bh_returns:
        fig_year.add_trace(go.Bar(x=x, y=[100*v for v in bh_returns], name="BH Return (%)", marker_color="#ff7f0e"))
    fig_year.update_layout(barmode="group", title="Yearly Returns: PF vs BH (percent %)", xaxis_title="Year", yaxis_title="Return (%)")
    if vbt_plot_width is not None:
        fig_year.update_layout(width=int(vbt_plot_width), height=int(fig_height*0.25))
    figs.append(fig_year)

    # ----- Stampa report se richiesto -----
    if show_report:
        try:
            my_display(df_summary, title=f"{pf.name if hasattr(pf,'name') else 'Portfolio'} - Timing Summary")
            my_display(df_yearly, title=f"{pf.name if hasattr(pf,'name') else 'Portfolio'} - Yearly Timing Breakdown")
            my_display(df_bh_vs_pf, title=f"{pf.name if hasattr(pf,'name') else 'Portfolio'} - Buy&Hold Comparator")
            my_display(df_timing, title=f"{pf.name if hasattr(pf,'name') else 'Portfolio'} - Timing Efficiency")
        except Exception:
            # fallback: print
            display(df_summary); display(df_yearly); display(df_bh_vs_pf); display(df_timing)

    aux = {
        "equity": eq,
        "returns": rets,
        "gross_exposure": ge,
        "invested_mask": invested_mask,
        "flat_mask": flat_mask,
        "partial_mask": partial_mask,
        "bh_series": bh_series
    }

    return {
        "df_summary": df_summary,
        "df_yearly": df_yearly,
        "df_bh_vs_pf": df_bh_vs_pf,
        "df_timing": df_timing,
        "aux": aux,
        "portfolio_type": portfolio_type,
        "figs": figs
    }
```

## Advanced Metrics

```python
# advanced_metrics.py

# ============================================================
# Utility di base
# ============================================================

def normalize_equity(equity: pd.Series) -> pd.Series:
    """
    Normalizza l'equity line:
      - ordina per data
      - rimuove NaN
      - rebasing a 1.0 sul primo valore
    """
    eq = equity.copy().dropna()
    eq = eq.sort_index()
    if len(eq) == 0:
        raise ValueError("Equity vuota dopo il dropna.")
    return eq / float(eq.iloc[0])


def equity_to_returns(equity: pd.Series, log: bool = False) -> pd.Series:
    """
    Converte una equity line normalizzata in una serie di rendimenti.
    """
    eq = normalize_equity(equity)
    if log:
        ret = np.log(eq / eq.shift(1))
    else:
        ret = eq.pct_change()
    return ret.fillna(0.0)


# ============================================================
# Metriche globali (point-in-time)
# ============================================================

def compute_cagr(equity: pd.Series, periods_per_year: int = 252) -> float:
    eq = normalize_equity(equity)
    n_periods = len(eq) - 1
    if n_periods <= 0:
        return np.nan
    years = n_periods / float(periods_per_year)
    final_val = float(eq.iloc[-1])
    if final_val <= 0 or years <= 0:
        return np.nan
    return final_val ** (1.0 / years) - 1.0


def compute_annualized_vol(returns: pd.Series, periods_per_year: int = 252) -> float:
    r = returns.dropna()
    if len(r) == 0:
        return np.nan
    return r.std(ddof=1) * np.sqrt(periods_per_year)


def compute_sharpe(
    returns: pd.Series,
    rf_rate_annual: float = 0.0,
    periods_per_year: int = 252
) -> float:
    r = returns.dropna()
    if len(r) == 0:
        return np.nan
    rf_periodic = rf_rate_annual / periods_per_year
    excess = r - rf_periodic
    vol = excess.std(ddof=1)
    if vol == 0 or np.isnan(vol):
        return np.nan
    return np.sqrt(periods_per_year) * excess.mean() / vol


def compute_sortino(
    returns: pd.Series,
    rf_rate_annual: float = 0.0,
    periods_per_year: int = 252
) -> float:
    r = returns.dropna()
    if len(r) == 0:
        return np.nan
    rf_periodic = rf_rate_annual / periods_per_year
    excess = r - rf_periodic
    downside = excess[excess < 0]
    if len(downside) == 0:
        return np.nan
    downside_std = downside.std(ddof=1)
    if downside_std == 0 or np.isnan(downside_std):
        return np.nan
    return np.sqrt(periods_per_year) * excess.mean() / downside_std


def compute_drawdown_stats(equity: pd.Series) -> dict:
    """
    Calcola:
      - max_drawdown: drawdown massimo (in %)
      - max_dd_duration: durata massima drawdown (in giorni)
      - dd_series: serie dei drawdown in %
    """
    eq = normalize_equity(equity)
    running_max = eq.cummax()
    dd = (eq / running_max) - 1.0  # valori <= 0

    max_dd = dd.min() if len(dd) > 0 else np.nan

    # durata massima del drawdown
    dd_is_zero = dd == 0
    max_dd_duration = 0
    current = 0
    for is_zero in dd_is_zero:
        if is_zero:
            current = 0
        else:
            current += 1
            max_dd_duration = max(max_dd_duration, current)

    return {
        "max_drawdown": float(max_dd),
        "max_dd_duration": int(max_dd_duration),
        "dd_series": dd
    }


def compute_ulcer_index(equity: pd.Series) -> float:
    """
    Ulcer Index: radice quadrata della media dei quadrati dei drawdown (%).
    """
    eq = normalize_equity(equity)
    running_max = eq.cummax()
    dd_pct = 100.0 * (eq / running_max - 1.0)  # in percentuale
    dd_sq = dd_pct.pow(2)
    if len(dd_sq) == 0:
        return np.nan
    return np.sqrt(dd_sq.mean())


def compute_calmar(cagr: float, max_drawdown: float) -> float:
    """
    Calmar ratio = CAGR / |MaxDD|
    MaxDD espresso come valore negativo (es. -0.3).
    """
    if max_drawdown >= 0 or np.isnan(max_drawdown):
        return np.nan
    return cagr / abs(max_drawdown)


def compute_point_in_time_stats(
    equity: pd.Series,
    rf_rate_annual: float = 0.0,
    periods_per_year: int = 252
) -> dict:
    """
    Metriche globali sull'intera equity line.
    """
    eq = normalize_equity(equity)
    ret = equity_to_returns(eq, log=False)

    cagr = compute_cagr(eq, periods_per_year=periods_per_year)
    vol = compute_annualized_vol(ret, periods_per_year=periods_per_year)
    sharpe = compute_sharpe(ret, rf_rate_annual=rf_rate_annual, periods_per_year=periods_per_year)
    sortino = compute_sortino(ret, rf_rate_annual=rf_rate_annual, periods_per_year=periods_per_year)

    dd_stats = compute_drawdown_stats(eq)
    ulcer = compute_ulcer_index(eq)
    calmar = compute_calmar(cagr, dd_stats["max_drawdown"])

    return {
        "CAGR": cagr,
        "Volatility": vol,
        "Sharpe": sharpe,
        "Sortino": sortino,
        "MaxDrawdown": dd_stats["max_drawdown"],
        "MaxDD_Duration": dd_stats["max_dd_duration"],
        "UlcerIndex": ulcer,
        "Calmar": calmar,
        "FinalValue": float(eq.iloc[-1]),
        "StartDate": eq.index[0],
        "EndDate": eq.index[-1],
        "N_Periods": len(eq)
    }


# ============================================================
# Analisi per finestre mobili (worst case 3/5/10 anni)
# ============================================================

def rolling_window_stats(
    equity: pd.Series,
    window_years: int = 3,
    periods_per_year: int = 252
) -> pd.DataFrame:
    """
    Per ogni finestra di lunghezza ~ window_years (in numero di barre),
    calcola metriche tipo:
      - start_date, end_date
      - length_years
      - CAGR_window
      - MaxDD_window
      - Ulcer_window
      - Vol_window
      - Sharpe_window
    """
    eq = normalize_equity(equity)
    n_periods = len(eq)
    window_size = int(window_years * periods_per_year)

    records = []
    for start_idx in range(0, n_periods - window_size + 1):
        end_idx = start_idx + window_size - 1
        eq_win = eq.iloc[start_idx:end_idx + 1]

        if len(eq_win) < 2:
            continue

        # metriche finestra
        ret_win = equity_to_returns(eq_win, log=False)
        cagr_win = compute_cagr(eq_win, periods_per_year=periods_per_year)
        vol_win = compute_annualized_vol(ret_win, periods_per_year=periods_per_year)
        sharpe_win = compute_sharpe(ret_win, rf_rate_annual=0.0, periods_per_year=periods_per_year)
        dd_stats_win = compute_drawdown_stats(eq_win)
        ulcer_win = compute_ulcer_index(eq_win)

        years_len = (len(eq_win) - 1) / float(periods_per_year)

        records.append({
            "start_date": eq_win.index[0],
            "end_date": eq_win.index[-1],
            "length_years": years_len,
            "CAGR_window": cagr_win,
            "Vol_window": vol_win,
            "Sharpe_window": sharpe_win,
            "MaxDD_window": dd_stats_win["max_drawdown"],
            "Ulcer_window": ulcer_win
        })

    if not records:
        return pd.DataFrame()

    df = pd.DataFrame(records)
    df.set_index("start_date", inplace=True)
    return df


def extract_worst_windows(
    df_rolling: pd.DataFrame,
    top_n: int = 5
) -> dict:
    """
    Estrae le peggiori finestre per:
      - CAGR_window (più basso)
      - MaxDD_window (drawdown più profondo)
      - Ulcer_window (più alto)
    """
    if df_rolling is None or df_rolling.empty:
        return {
            "worst_by_cagr": pd.DataFrame(),
            "worst_by_maxdd": pd.DataFrame(),
            "worst_by_ulcer": pd.DataFrame()
        }

    worst_by_cagr = df_rolling.sort_values("CAGR_window").head(top_n)
    worst_by_maxdd = df_rolling.sort_values("MaxDD_window").head(top_n)  # più negativo = peggiore
    worst_by_ulcer = df_rolling.sort_values("Ulcer_window", ascending=False).head(top_n)

    return {
        "worst_by_cagr": worst_by_cagr,
        "worst_by_maxdd": worst_by_maxdd,
        "worst_by_ulcer": worst_by_ulcer
    }


# ============================================================
# Funzione “front-end” del modulo
# ============================================================

def compute_advanced_metrics(
    equity: pd.Series,
    rf_rate_annual: float = 0.0,
    periods_per_year: int = 252,
    window_years_list: tuple[int, ...] = (3, 5, 10),
    top_n_worst: int = 5
) -> dict:
    """
    Wrapper unico per il framework.

    Ritorna un dict con:
      - 'point_in_time': metriche globali
      - 'rolling_windows': {anni: df_rolling}
      - 'worst_windows': {anni: {'worst_by_cagr', 'worst_by_maxdd', 'worst_by_ulcer'}}
    """
    eq = normalize_equity(equity)

    point_stats = compute_point_in_time_stats(
        eq,
        rf_rate_annual=rf_rate_annual,
        periods_per_year=periods_per_year
    )

    rolling_dict = {}
    worst_dict = {}

    for wy in window_years_list:
        df_roll = rolling_window_stats(
            eq,
            window_years=wy,
            periods_per_year=periods_per_year
        )
        rolling_dict[wy] = df_roll
        worst_dict[wy] = extract_worst_windows(df_roll, top_n=top_n_worst)

    return {
        "point_in_time": point_stats,
        "rolling_windows": rolling_dict,
        "worst_windows": worst_dict
    }

def extract_equity_series(portfolio_obj) -> pd.Series:
    """
    Prova a estrarre un'equity line (pd.Series) da un oggetto portafoglio
    (vectorbt.Portfolio, DataFrame o Series).
    """
    import pandas as pd

    # Caso vectorbt.Portfolio (ha il metodo value)
    if hasattr(portfolio_obj, "value"):
        val = portfolio_obj.value()
    else:
        val = portfolio_obj

    if isinstance(val, pd.DataFrame):
        # usa la prima colonna
        return val.iloc[:, 0]
    elif isinstance(val, pd.Series):
        return val
    else:
        raise TypeError("Tipo di portfolio non riconosciuto per l'estrazione dell'equity.")
import pandas as pd
import numpy as np
from IPython.display import display

# Se il file è in un modulo separato:
# from advanced_metrics import extract_equity_series, compute_advanced_metrics

# ------------------------------------------------------------
# Helper: formattazione/printing in Jupyter
# ------------------------------------------------------------
def _format_point_stats(stats: dict) -> pd.DataFrame:
    """Trasforma il dict point_in_time in tabella 2 colonne (Metric, Value)."""
    df = pd.DataFrame(stats, index=[0]).T
    df.columns = ["Value"]
    df.index.name = "Metric"
    return df

def _style_point_stats(df: pd.DataFrame) -> "pd.io.formats.style.Styler":
    """Stile: percentuali e date leggibili."""
    pct_rows = {"CAGR", "Volatility", "MaxDrawdown"}
    num_rows = {"Sharpe", "Sortino", "UlcerIndex", "Calmar", "FinalValue"}
    date_rows = {"StartDate", "EndDate"}
    int_rows = {"N_Periods", "MaxDD_Duration"}

    def fmt(v, idx):
        if pd.isna(v):
            return ""
        if idx in pct_rows:
            return f"{float(v)*100:.2f}%"
        if idx in num_rows:
            # FinalValue lo stampo con separatore, gli altri con 2 decimali
            if idx == "FinalValue":
                return f"{float(v):,.2f}"
            return f"{float(v):.2f}"
        if idx in date_rows:
            return pd.to_datetime(v).strftime("%Y-%m-%d")
        if idx in int_rows:
            return f"{int(v)}"
        return str(v)

    styled = df.copy()
    styled["Value"] = [fmt(df.loc[i, "Value"], i) for i in df.index]
    return styled.style.set_properties(**{"white-space": "nowrap"}) \
                     .set_table_styles([{"selector": "th", "props": [("text-align", "left")]}])

def _style_worst_windows(df: pd.DataFrame) -> "pd.io.formats.style.Styler":
    """Stile per le tabelle worst windows."""
    if df is None or df.empty:
        return pd.DataFrame().style

    def pct(x):
        return "" if pd.isna(x) else f"{float(x)*100:.2f}%"

    def num(x):
        return "" if pd.isna(x) else f"{float(x):.2f}"

    def date(x):
        return "" if pd.isna(x) else pd.to_datetime(x).strftime("%Y-%m-%d")

    fmt_map = {
        "end_date": date,
        "CAGR_window": pct,
        "Vol_window": pct,
        "Sharpe_window": num,
        "MaxDD_window": pct,      # è negativo (drawdown)
        "Ulcer_window": num,
        "length_years": num,
    }

    cols = [c for c in ["end_date","length_years","CAGR_window","Vol_window","Sharpe_window","MaxDD_window","Ulcer_window"] if c in df.columns]
    out = df[cols].copy()

    return out.style.format(fmt_map) \
                   .set_properties(**{"white-space": "nowrap"}) \
                   .set_table_styles([{"selector": "th", "props": [("text-align", "left")]}])

# ------------------------------------------------------------
# Front-end: stampa completa in Jupyter
# ------------------------------------------------------------
def show_advanced_metrics_in_notebook(
    portfolio_obj,
    rf_rate_annual: float = 0.01,
    periods_per_year: int = 252,
    window_years_list: tuple[int, ...] = (3, 5, 10),
    top_n_worst: int = 5,
    title: str | None = None,
):
    """
    Stampa in Jupyter:
      - metriche globali
      - worst windows per 3/5/10 anni (CAGR, MaxDD, Ulcer)
    """
    # 1) equity series
    equity = extract_equity_series(portfolio_obj)

    # 2) calcolo metriche
    adv = compute_advanced_metrics(
        equity=equity,
        rf_rate_annual=rf_rate_annual,
        periods_per_year=periods_per_year,
        window_years_list=window_years_list,
        top_n_worst=top_n_worst
    )

    # 3) stampa
    if title:
        print(title)

    # --- Point-in-time ---
    pt_df = _format_point_stats(adv["point_in_time"])
    display(_style_point_stats(pt_df))

    # --- Worst windows ---
    for wy in window_years_list:
        w = adv["worst_windows"].get(wy, {})
        print(f"\nWorst windows ({wy} anni) — top {top_n_worst}")

        print("• Worst by CAGR (più basso)")
        display(_style_worst_windows(w.get("worst_by_cagr")))

        print("• Worst by Max Drawdown (più profondo)")
        display(_style_worst_windows(w.get("worst_by_maxdd")))

        print("• Worst by Ulcer (più alto)")
        display(_style_worst_windows(w.get("worst_by_ulcer")))

    return adv  # utile se vuoi riusarlo (salvare/csv/report)

def _compute_drawdown_series(equity: pd.Series) -> pd.Series:
    eq = normalize_equity(equity)
    peak = eq.cummax()
    dd = (eq / peak) - 1.0
    return dd

def _rolling_cagr(equity: pd.Series, window_years: int = 3, periods_per_year: int = 252) -> pd.Series:
    """
    Rolling CAGR su finestre di window_years (approssimate in barre).
    """
    eq = normalize_equity(equity)
    w = int(window_years * periods_per_year)
    if len(eq) < w + 1:
        return pd.Series(index=eq.index, dtype=float)

    # CAGR = (eq_t / eq_{t-w}) ** (1/years) - 1
    years = w / float(periods_per_year)
    cagr = (eq / eq.shift(w)) ** (1.0 / years) - 1.0
    return cagr

def _plot_advanced_diagnostics(
    equity: pd.Series,
    periods_per_year: int = 252,
    rolling_years_list: tuple[int, ...] = (3, 5, 10),
    log_equity: bool = True,
    title: str | None = None
):
    eq = normalize_equity(equity)
    dd = _compute_drawdown_series(eq)

    # --- Plot 1: Equity ---
    plt.figure()
    if title:
        plt.title(title + " — Equity (normalized)")
    else:
        plt.title("Equity (normalized)")
    plt.plot(eq.index, eq.values)
    plt.yscale("log" if log_equity else "linear")
    plt.grid(True, alpha=0.3)
    plt.show()

    # --- Plot 2: Drawdown ---
    plt.figure()
    if title:
        plt.title(title + " — Drawdown")
    else:
        plt.title("Drawdown")
    plt.plot(dd.index, dd.values)
    plt.axhline(0.0, linewidth=1)
    plt.grid(True, alpha=0.3)
    plt.show()

    # --- Plot 3: Rolling CAGR (più curve, una per finestra) ---
    plt.figure()
    if title:
        plt.title(title + " — Rolling CAGR")
    else:
        plt.title("Rolling CAGR")

    any_plotted = False
    for wy in rolling_years_list:
        rc = _rolling_cagr(eq, window_years=wy, periods_per_year=periods_per_year)
        if rc.notna().sum() > 0:
            plt.plot(rc.index, rc.values, label=f"{wy}y")
            any_plotted = True

    if any_plotted:
        plt.axhline(0.0, linewidth=1)
        plt.legend()
        plt.grid(True, alpha=0.3)
        plt.show()
    else:
        plt.close()
        print("Rolling CAGR: storico insufficiente per le finestre richieste.")

def show_advanced_metrics_in_notebook_with_plots(
    portfolio_obj,
    rf_rate_annual: float = 0.01,
    periods_per_year: int = 252,
    window_years_list: tuple[int, ...] = (3, 5, 10),
    top_n_worst: int = 5,
    title: str | None = None,
    # --- plots ---
    plots: bool = True,
    log_equity: bool = True,
):
    """
    Stampa tabelle + (opzionale) grafici diagnostici.
    """
    # 1) equity
    equity = extract_equity_series(portfolio_obj)

    # 2) metriche (tabelle)
    adv = compute_advanced_metrics(
        equity=equity,
        rf_rate_annual=rf_rate_annual,
        periods_per_year=periods_per_year,
        window_years_list=window_years_list,
        top_n_worst=top_n_worst
    )

    if title:
        print(title)

    # --- Point-in-time table ---
    pt_df = pd.DataFrame(adv["point_in_time"], index=[0]).T
    pt_df.columns = ["Value"]
    pt_df.index.name = "Metric"
    display(pt_df)

    # --- Worst windows tables ---
    for wy in window_years_list:
        w = adv["worst_windows"].get(wy, {})
        print(f"\nWorst windows ({wy} anni) — top {top_n_worst}")
        print("• Worst by CAGR (più basso)")
        display(w.get("worst_by_cagr", pd.DataFrame()))
        print("• Worst by Max Drawdown (più profondo)")
        display(w.get("worst_by_maxdd", pd.DataFrame()))
        print("• Worst by Ulcer (più alto)")
        display(w.get("worst_by_ulcer", pd.DataFrame()))

    # 3) plots
    if plots:
        _plot_advanced_diagnostics(
            equity=equity,
            periods_per_year=periods_per_year,
            rolling_years_list=window_years_list,
            log_equity=log_equity,
            title=title
        )

    return adv

# # Uso:
equity_series    = extract_equity_series(portfolio)
# # bh_equity_series = extract_equity_series(bh_portfolio)

adv_metrics    = compute_advanced_metrics(equity_series, rf_rate_annual=0.01)
# # adv_metrics_bh = compute_advanced_metrics(bh_equity_series, rf_rate_annual=0.01)
```

## Mister K Code Example

```python
# Codice Mister K corretto

import numpy as np
import pandas as pd
import yfinance as yf
import vectorbt as vbt
import itertools

# ========================== Indicator Functions ==========================

def calculate_cog(df, period):
    weighted_sum = sum((i + 1) * df['Close'].shift(i) for i in range(period))
    cog = weighted_sum / sum(range(1, period + 1))
    return cog

def calculate_real_qqe_mod(df, rsi_period, smoothing_factor, wilders_period):
    close = df['Close']
    delta = close.diff()
    gain = delta.where(delta > 0, 0.0)
    loss = -delta.where(delta < 0, 0.0)

    avg_gain = gain.ewm(alpha=1/rsi_period, adjust=False, min_periods=1).mean()
    avg_loss = loss.ewm(alpha=1/rsi_period, adjust=False, min_periods=1).mean().replace(0, np.nan)
    rs = avg_gain / avg_loss
    rsi = 100 - (100 / (1 + rs))

    smoothed_rsi = rsi.ewm(span=smoothing_factor, adjust=False, min_periods=1).mean()
    rsi_tr = smoothed_rsi.diff().abs()
    atr_rsi = rsi_tr.ewm(span=wilders_period, adjust=False, min_periods=1).mean()

    d_factor = 4.236
    qqe_trailing = smoothed_rsi - (atr_rsi * d_factor)
    signal_line = qqe_trailing.ewm(span=smoothing_factor, adjust=False, min_periods=1).mean()

    return smoothed_rsi, signal_line, atr_rsi

# ========================== Walk-Forward Optimization ==========================
from tqdm.auto import tqdm  # <-- aggiungi questo import

def walk_forward_optimization_cog_qqe(df, start_year, end_year, show_progress=True):
    results = []

    # griglie parametri
    cog_range       = range(5, 20, 2)
    rsi_range       = range(8, 21, 2)
    smoothing_range = range(3, 11)
    wilders_range   = range(8, 21, 2)
    shift_range     = range(10, 31, 2)

    # prodotto cartesiano e contatore totale dei casi
    combos = list(itertools.product(cog_range, rsi_range, smoothing_range, wilders_range, shift_range))
    n_combos = len(combos)
    test_years = list(range(start_year + 4, end_year + 1))
    total_cases = n_combos * len(test_years)

    pbar = tqdm(total=total_cases, disable=not show_progress, desc="WFO grid cases")

    for test_year in test_years:
        train_start = test_year - 4
        train_end   = test_year - 1

        train_data = df[(df.index.year >= train_start) & (df.index.year <= train_end)]
        test_data  = df[df.index.year == test_year]

        best_params = None
        best_performance = -np.inf

        # loop sulle combinazioni con aggiornamento della barra
        for params in combos:
            cog_p, rsi_p, smooth_p, wilders_p, shift = params
            temp = train_data.copy()
            temp['COG'] = calculate_cog(temp, cog_p)
            temp['QQE'], temp['QQE_Signal'], temp['ATR_RSI'] = calculate_real_qqe_mod(
                temp, rsi_p, smooth_p, wilders_p
            )

            entries = (temp['Close'] < temp['COG']) & (temp['QQE'] > temp['QQE'].shift(shift))
            exits   = (temp['Close'] > temp['COG']) & (temp['QQE'] < temp['QQE'].shift(shift))

            entries = entries.shift(1).astype(bool).fillna(False).to_numpy()
            exits   = exits.shift(1).astype(bool).fillna(False).to_numpy()

            portfolio = vbt.Portfolio.from_signals(
                close=temp['Open'],
                entries=entries,
                exits=exits,
                init_cash=100_000,
                fees=0.001,
                slippage=0.002,
                freq='D'
            )

            perf = portfolio.total_return()
            if perf > best_performance:
                best_performance = perf
                best_params = params

            pbar.update(1)                 # <-- avanzamento per ogni combinazione
            pbar.set_postfix(year=test_year)

        # applica best_params all'anno di test (come in origine)
        test_year_data = df[(df.index.year >= test_year - 1) & (df.index.year <= test_year)].copy()
        cog_p, rsi_p, smooth_p, wilders_p, shift = best_params
        test_year_data['COG'] = calculate_cog(test_year_data, cog_p)
        test_year_data['QQE'], test_year_data['QQE_Signal'], test_year_data['ATR_RSI'] = calculate_real_qqe_mod(
            test_year_data, rsi_p, smooth_p, wilders_p
        )
        test_year_data = test_year_data[test_year_data.index.year == test_year]

        entries = (test_year_data['Close'] < test_year_data['COG']) & (test_year_data['QQE'] > test_year_data['QQE'].shift(shift))
        exits   = (test_year_data['Close'] > test_year_data['COG']) & (test_year_data['QQE'] < test_year_data['QQE'].shift(shift))

        entries = entries.shift(1).astype(bool).fillna(False).to_numpy()
        exits   = exits.shift(1).astype(bool).fillna(False).to_numpy()

        portfolio = vbt.Portfolio.from_signals(
            close=test_year_data['Open'],
            entries=entries,
            exits=exits,
            init_cash=100_000,
            fees=0.001,
            slippage=0.002,
            freq='D'
        )

        results.append({
            'Year': test_year,
            'Best_Params': best_params,
            'Test_Return': portfolio.total_return()
        })

    pbar.close()
    return pd.DataFrame(results)

# ========================== Run the Optimization ==========================

symbol = 'TPR'
start_date = '2015-01-01'
end_date = '2025-01-01'

df = yf.download(symbol, start=start_date, end=end_date, multi_level_index=False)
results = walk_forward_optimization_cog_qqe(df, 2016, 2025)

print("\nWalk-Forward Optimization Results:")
print(results)

# ========================== Combine All Signals ==========================

combined_entries = pd.Series(False, index=df.index)
combined_exits = pd.Series(False, index=df.index)

for _, row in results.iterrows():
    year = row['Year']
    cog_p, rsi_p, smooth_p, wilders_p, shift = row['Best_Params']

    yearly_data = df[(df.index.year >= year - 1) & (df.index.year <= year)].copy()
    yearly_data['COG'] = calculate_cog(yearly_data, cog_p)
    yearly_data['QQE'], yearly_data['QQE_Signal'], yearly_data['ATR_RSI'] = calculate_real_qqe_mod(yearly_data, rsi_p, smooth_p, wilders_p)

    yearly_data = yearly_data[yearly_data.index.year == year]

    entries = (yearly_data['Close'] < yearly_data['COG']) & (yearly_data['QQE'] > yearly_data['QQE'].shift(shift))
    exits = (yearly_data['Close'] > yearly_data['COG']) & (yearly_data['QQE'] < yearly_data['QQE'].shift(shift))

    combined_entries.loc[entries.index] = combined_entries.loc[entries.index] | entries
    combined_exits.loc[exits.index] = combined_exits.loc[exits.index] | exits

# Final backtest on test period
df = df[(df.index.year >= 2020) & (df.index.year <= 2025)]
combined_entries = combined_entries[(combined_entries.index.year >= 2020) & (combined_entries.index.year <= 2025)]
combined_exits = combined_exits[(combined_exits.index.year >= 2020) & (combined_exits.index.year <= 2025)]

# Shift entries and exits to next bar
shifted_entries = combined_entries.shift(1).astype(bool).fillna(False).to_numpy()
shifted_exits = combined_exits.shift(1).astype(bool).fillna(False).to_numpy()

# Create portfolio using 'Open' prices
portfolio = vbt.Portfolio.from_signals(
    close=df['Open'],
    entries=shifted_entries,
    exits=shifted_exits,
    init_cash=100_000,
    fees=0.001,
    slippage=0.002,
    freq='D'
)

# Parameters
risk_free_rate = 0.02  # Annual risk-free rate, e.g., 2%

# Get core metrics
car = portfolio.annualized_return()
vol = portfolio.annualized_volatility()
max_dd = abs(portfolio.max_drawdown())

# Adjusted CAR using Sharpe-style formula (based on volatility)
adj_car_sharpe = (car - risk_free_rate) / vol if vol != 0 else np.nan

# Adjusted CAR using Calmar-style formula (based on drawdown)
adj_car_calmar = car / max_dd if max_dd != 0 else np.nan

# Initialize variables to track exposure calculation
in_position = False
exposure_days = 0
total_days = len(df)

# Loop through the entries and exits to calculate exposure
for i in range(1, len(df)):
    if combined_entries[i] and not in_position:  # Entry signal (enter position)
        in_position = True
        entry_day = i  # Track entry day
    elif combined_exits[i] and in_position:  # Exit signal (exit position)
        in_position = False
        exit_day = i  # Track exit day
        exposure_days += exit_day - entry_day  # Count days in position

# If the last position is still open (i.e., no exit signal), consider the last day of the dataset
if in_position:
    exposure_days += total_days - entry_day  # Count remaining days in position

# Calculate exposure as percentage of time in position
exposure_percentage = (exposure_days / total_days) * 100

stats = portfolio.stats()
stats['CAR'] = f"{car:.2%}"
stats['Adjusted CAR (Sharpe-style)'] = f"{adj_car_sharpe:.2%}"
stats['Adjusted CAR (Calmar-style)'] = f"{adj_car_calmar:.2%}"
stats['Market Time Exposure'] = f"{exposure_percentage:.2f}%"

# Display performance metrics
print(stats)

# Plot equity curve
portfolio.plot().show()
```
